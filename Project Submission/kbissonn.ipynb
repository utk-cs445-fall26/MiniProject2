```python
# Script that looks at df_commit_data.csv to pull the number of commits, number of authors, and min/max time

import pandas as pd
from datetime import datetime, timezone

df = pd.read_csv('df_commit_data.csv')

# group by project and aggregate
summary = df.groupby('project').agg(
    num_commits=('commit', 'nunique'),
    num_authors=('author', 'nunique'),
    max_time=('author_time', 'max'),
    min_time=('author_time', 'min'),
).reset_index()

# convert unix timestamps to readable datetimes
summary['max_time_readable'] = summary['max_time'].apply(
    lambda t: datetime.fromtimestamp(t, tz=timezone.utc).isoformat()
)
summary['min_time_readable'] = summary['min_time'].apply(
    lambda t: datetime.fromtimestamp(t, tz=timezone.utc).isoformat()
)

for _, row in summary.iterrows():
    print(f"{row['project']}: commits={row['num_commits']}, authors={row['num_authors']}, "
          f"max_time={row['max_time_readable']}, min_time={row['min_time_readable']}")
```

```python
# Part 3, pulling the commits, see kbissonn_poject_gap_commits.csv for results
import pandas as pd

N = 10
IN_FILE = "kbissonn_project_summary.csv"
OUT_FILE = "kbissonn_project_gap_commits.csv"

df = pd.read_csv(IN_FILE, sep=";", dtype=str, keep_default_na=False)
orig_cols = ["project_wocid", "commit_sha1", "author", "time", "commit_message"]
df.columns = orig_cols  # rename by position

# time may be a unix timestamp (World of Code) or a date string
t = pd.to_numeric(df["time"], errors="coerce")
if t.notna().all():
    df["_dt"] = pd.to_datetime(t, unit="s", utc=True)
else:
    df["_dt"] = pd.to_datetime(df["time"], utc=True, errors="coerce")
df = df.dropna(subset=["_dt"])
df["_month"] = df["_dt"].dt.tz_localize(None).dt.to_period("M")


def longest_gap(d):
    """Return (gap_start, gap_end, length) of the longest run of 0-commit months."""
    counts = d.groupby("_month").size()
    idx = pd.period_range(counts.index.min(), counts.index.max(), freq="M")
    counts = counts.reindex(idx, fill_value=0)
    best = (None, None, 0)
    start = None
    for m, c in counts.items():
        if c == 0:
            if start is None:
                start = m
            length = (m - start).n + 1
            if length > best[2]:
                best = (start, m, length)
        else:
            start = None
    return best


parts = []
for project, d in df.groupby("project_wocid"):
    d = d.sort_values("_dt")
    gs, ge, glen = longest_gap(d)
    if glen == 0:
        print(f"{project}: no gap found, skipped")
        continue

    pre = d[d["_month"] < gs].tail(N)
    post = d[d["_month"] > ge].head(N)
    print(f"{project}: gap {gs} to {ge} ({glen} months), "
          f"{len(pre)} pre, {len(post)} post")

    parts.append(pre.assign(pre_post="pre"))
    parts.append(post.assign(pre_post="post"))

out = pd.concat(parts)[["pre_post"] + orig_cols]
out.to_csv(OUT_FILE, sep=";", index=False)
```
