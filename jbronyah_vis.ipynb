```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker
```

```python
df = pd.read_csv("jbronyah_project_summary.csv", sep=";")
df.columns = ['project','commit','author','time','message']
df['Month'] = pd.to_datetime(df['time'], unit='s').dt.strftime('%Y-%m')
df.head()
```

```python
netid = 'jbronyah'

projects = [
    'pecanproject_pecan',
    'esa-philab_opensartoolkit',
    'stolllab_easyspin',
    'mariohsouto_proxsdp.jl',
    'qiskit_qiskit-experiments',
    'ericproffitt_topicmodelsvb.jl',
    'vcityteam_ud-reproducibility',
    'hbctraining_scrna-seq_online',
    'smileipic_smilei',
    'geofranzi_geoportal',
]

all_timeseries = []
gap_results = []

for prj in projects:
  # Filter project and aggregate monthly counts
  df1 = df[df['project'] == prj]
  if df1.empty:
    continue
  df2 = df1.groupby('Month').size().reset_index(name='Commits')
  monthly_counts = df2
  monthly_counts['time'] = pd.to_datetime(monthly_counts['Month'])
  full_date_range = pd.DataFrame({
      'time': pd.date_range(
          start=monthly_counts['time'].min(),
          end=monthly_counts['time'].max(),
          freq='MS',
      )
  })
  monthly_counts = pd.merge(
      full_date_range, monthly_counts, on='time', how='left'
  ).fillna(0)
  monthly_counts['Month'] = monthly_counts['time'].dt.strftime('%Y-%m')
  monthly_counts['project'] = prj

  # -----------------------------------------------------------------
  # Collect & save timeseries data
  # -----------------------------------------------------------------
  all_timeseries.append(monthly_counts[['project', 'Month', 'Commits']])
  monthly_counts[['project', 'Month', 'Commits']].to_csv(
      f'{netid}_commits_timeseries_{prj}.csv', sep=';', index=False
  )

  # -----------------------------------------------------------------
  # Plotting
  # -----------------------------------------------------------------
  plt.figure(figsize=(10, 5))
  plt.plot(
      monthly_counts['Month'],
      monthly_counts['Commits'],
      marker='o',
      linestyle='-',
  )
  plt.xlabel('Time (YYYY-MM)')
  plt.ylabel('Number of Commits')
  plt.title('Commit Activity: ' + prj)
  plt.grid(True)
  plt.tight_layout()
  ax = plt.gca()
  for label in ax.xaxis.get_ticklabels()[::2]:
    label.set_visible(False)
  plt.xticks(rotation=45, fontsize=7)
  plt.savefig(f'{netid}_timeseries_' + prj + '.png', dpi=600)
  plt.show()
  plt.close()

  # -----------------------------------------------------------------
  # Analyze Inactivity Gaps
  # -----------------------------------------------------------------
  zero_runs = []
  start_idx = None
  cur_len = 0

  for i, row in monthly_counts.iterrows():
    if row['Commits'] == 0:
      if start_idx is None:
        start_idx = i
      cur_len += 1
    else:
      if start_idx is not None:
        zero_runs.append((start_idx, i - 1, cur_len))
        start_idx = None
        cur_len = 0
  if start_idx is not None:
    zero_runs.append((start_idx, len(monthly_counts) - 1, cur_len))

  if zero_runs:
    longest = max(zero_runs, key=lambda x: x[2])
    gap_start_idx, gap_end_idx, gap_len = longest
    gap_start = monthly_counts.loc[gap_start_idx, 'Month']
    gap_end = monthly_counts.loc[gap_end_idx, 'Month']
    commits_after = (
        int(monthly_counts.loc[gap_end_idx + 1 :, 'Commits'].sum())
        if gap_end_idx + 1 < len(monthly_counts)
        else 0
    )
  else:
    gap_start = 'None'
    gap_end = 'None'
    gap_len = 0
    commits_after = 0

  # Activity pattern classification heuristic
  mid = len(monthly_counts) // 2
  first_half = monthly_counts.loc[:mid, 'Commits'].sum()
  second_half = monthly_counts.loc[mid:, 'Commits'].sum()
  last_6_months = monthly_counts.loc[
      max(0, len(monthly_counts) - 6) :, 'Commits'
  ].sum()

  if last_6_months == 0 and commits_after < 5:
    pattern = 'dormant'
  elif second_half < first_half * 0.5:
    pattern = 'declining'
  elif second_half > first_half * 1.5:
    pattern = 'increasing'
  else:
    pattern = 'sporadic'

  gap_results.append({
      'project': prj,
      'LongestGapStart (YYYY-MM), ': gap_start,
      'LongestGapEnd (YYYY-MM),': gap_end,
      'LongestGapLength': gap_len,
      'NumCommitsAfterLastGap': commits_after,
      'pattern': pattern,
  })

# =====================================================================
# Export Consolidated Timeseries CSV
# =====================================================================
df_all_timeseries = pd.concat(all_timeseries, ignore_index=True)
df_all_timeseries.to_csv(f'{netid}_commits_timeseries.csv', sep=';', index=False)

# =====================================================================
# Update jbronyah_project_stats.csv With Gap Results
# =====================================================================
stats_filename = f'{netid}_project_stats.csv'
existing_stats = pd.read_csv(stats_filename, sep=';')

# Keep base 8 columns from Part 1
base_cols = [
    'project',
    '#commits',
    '#authors',
    'From',
    'To',
    '#stars',
    '#forks',
    'LastGHCommitDate',
]
existing_stats = existing_stats[
    [c for c in base_cols if c in existing_stats.columns]
]

# Merge Part 2 gap/pattern columns 
df_gaps = pd.DataFrame(gap_results)
updated_stats = existing_stats.merge(df_gaps, on='project', how='left')

# Save updated CSV
updated_stats.to_csv(stats_filename, sep=';', index=False)

print('Updated files:')
print(f'1. {netid}_commits_timeseries.csv')
print(f'2. {netid}_project_stats.csv (updated to all 13 columns)')
display(updated_stats)
```

```python
import os
import re
import numpy as np
import pandas as pd

netid = 'jbronyah'

# Load project summary and current stats
df_summary = pd.read_csv(f'{netid}_project_summary.csv', sep=';', header=None)
df_summary.columns = ['project', 'commit', 'author', 'time', 'message']
df_summary['datetime'] = pd.to_datetime(df_summary['time'], unit='s')
df_summary['Month'] = df_summary['datetime'].dt.strftime('%Y-%m')

stats_file = f'{netid}_project_stats.csv'
df_stats = pd.read_csv(stats_file, sep=';')


# Helper function to compute timeline gaps directly from commit history
def compute_project_gaps(df_prj):
  df_monthly = df_prj.groupby('Month').size().reset_index(name='Commits')
  df_monthly['time'] = pd.to_datetime(df_monthly['Month'])

  full_range = pd.DataFrame({
      'time': pd.date_range(
          start=df_monthly['time'].min(),
          end=df_monthly['time'].max(),
          freq='MS',
      )
  })
  monthly = pd.merge(full_range, df_monthly, on='time', how='left').fillna(0)
  monthly['Month'] = monthly['time'].dt.strftime('%Y-%m')

  zero_runs = []
  start_idx = None
  cur_len = 0
  for i, row in monthly.iterrows():
    if row['Commits'] == 0:
      if start_idx is None:
        start_idx = i
      cur_len += 1
    else:
      if start_idx is not None:
        zero_runs.append((start_idx, i - 1, cur_len))
        start_idx = None
        cur_len = 0
  if start_idx is not None:
    zero_runs.append((start_idx, len(monthly) - 1, cur_len))

  if zero_runs:
    longest = max(zero_runs, key=lambda x: x[2])
    gap_start_idx, gap_end_idx, gap_len = longest
    gap_start = monthly.loc[gap_start_idx, 'Month']
    gap_end = monthly.loc[gap_end_idx, 'Month']
    commits_after = (
        int(monthly.loc[gap_end_idx + 1 :, 'Commits'].sum())
        if gap_end_idx + 1 < len(monthly)
        else 0
    )
  else:
    gap_start = 'None'
    gap_end = 'None'
    gap_len = 0
    commits_after = 0

  mid = len(monthly) // 2
  first_half = monthly.loc[:mid, 'Commits'].sum()
  second_half = monthly.loc[mid:, 'Commits'].sum()
  last_6_months = monthly.loc[
      max(0, len(monthly) - 6) :, 'Commits'
  ].sum()

  if last_6_months == 0 and commits_after < 5:
    pattern = 'dormant'
  elif second_half < first_half * 0.5:
    pattern = 'declining'
  elif second_half > first_half * 1.5:
    pattern = 'increasing'
  else:
    pattern = 'sporadic'

  gaps_ge_3 = sum(1 for r in zero_runs if r[2] >= 3)

  return {
      'LongestGapStart': gap_start,
      'LongestGapEnd': gap_end,
      'LongestGapLength': gap_len,
      'NumCommitsAfterLastGap': commits_after,
      'ActivityPattern': pattern,
      'NumberOfGapsInTimeline': gaps_ge_3,
  }


# Helper function to classify commit messages into standard themes
def classify_message(msg):
  if not isinstance(msg, str):
    return 'Other'
  msg_lower = msg.lower()
  if any(w in msg_lower for w in ['fix', 'bug', 'patch', 'issue', 'resolve']):
    return 'Bug fixes'
  elif any(
      w in msg_lower
      for w in ['doc', 'readme', 'changelog', 'typo', 'manual', 'license']
  ):
    return 'Documentation updates'
  elif any(
      w in msg_lower
      for w in ['dep', 'bump', 'upgrade', 'compat', 'requirement', 'version']
  ):
    return 'Dependency updates'
  elif any(
      w in msg_lower
      for w in ['release', 'tag', 'v0.', 'v1.', 'v2.', 'publish']
  ):
    return 'Release'
  elif any(w in msg_lower for w in ['sec', 'cve', 'vulnerab', 'auth']):
    return 'Security patches'
  elif any(
      w in msg_lower
      for w in ['dependabot', 'bot', 'github-actions', 'ci', 'travis']
  ):
    return 'Automated bot contributions'
  elif any(
      w in msg_lower
      for w in ['add', 'feat', 'new', 'implement', 'support', 'create']
  ):
    return 'Feature development'
  else:
    return 'Other'


def get_top_two_themes(df_slice):
  if df_slice.empty:
    return 'Other:Other'
  themes = df_slice['message'].apply(classify_message)
  counts = themes.value_counts()
  top = list(counts.index[:2])
  if len(top) == 1:
    top.append('Other')
  return f'{top[0]}:{top[1]}'


# Extract gap commits and calculate Part 3.1 & 3.2 metrics
gap_commits_list = []
part3_data = []

cutoff_date = pd.to_datetime('2025-04-01')

for idx, row in df_stats.iterrows():
  prj = row['project']
  df_prj = df_summary[df_summary['project'] == prj].sort_values('time')
  if df_prj.empty:
    continue

  # Dynamically determine the longest gap
  gap_metrics = compute_project_gaps(df_prj)
  start_str = str(gap_metrics['LongestGapStart']).strip()
  end_str = str(gap_metrics['LongestGapEnd']).strip()

  # Safe validation: must match YYYY-MM format
  has_gap = bool(
      re.match(r'^\d{4}-\d{2}$', start_str)
      and re.match(r'^\d{4}-\d{2}$', end_str)
  )

  if has_gap:
    start_dt = pd.to_datetime(start_str + '-01')
    end_dt = (
        pd.to_datetime(end_str + '-01')
        + pd.offsets.MonthEnd(1)
        + pd.Timedelta(days=1)
    )

    pre_df = df_prj[df_prj['datetime'] < start_dt].tail(10).copy()
    post_df = df_prj[df_prj['datetime'] > end_dt].head(10).copy()
  else:
    pre_df = df_prj.tail(10).copy()
    post_df = pd.DataFrame(columns=df_prj.columns)

  pre_df['pre/post'] = 'pre'
  post_df['pre/post'] = 'post'

  combined_gap = pd.concat([pre_df, post_df])
  gap_commits_list.append(combined_gap)

  before_themes = get_top_two_themes(pre_df)
  after_themes = get_top_two_themes(post_df) if not post_df.empty else 'N/A:N/A'

  recent_df = df_prj.tail(10)
  recent_themes = get_top_two_themes(recent_df)

  pre_authors = set(pre_df['author'].dropna().unique())
  post_authors = (
      set(post_df['author'].dropna().unique()) if not post_df.empty else set()
  )
  overlap = pre_authors.intersection(post_authors)

  if not has_gap:
    hypo_reason = 'N/A active Project'
    has_recovered = 'N/A'
    why_recovered = 'N/A'
    who_recovered = 'N/A'
  elif post_df.empty:
    hypo_reason = 'Development ceased following stabilization or graduation'
    has_recovered = 'No'
    why_recovered = 'No recovery commits observed to date'
    who_recovered = 'N/A'
  else:
    hypo_reason = 'Feature freeze / maintenance pause after core milestone'
    has_recovered = 'Yes'
    why_recovered = (
        'Maintenance updates, bug fixes, or compatibility adjustments'
    )
    if len(overlap) > 0 and len(post_authors - overlap) > 0:
      who_recovered = 'Both original authors and new contributors'
    elif len(overlap) > 0:
      who_recovered = 'Original authors continued maintenance'
    else:
      who_recovered = 'New contributors joined to revive the project'

  last_gh_date_str = str(row.get('LastGHCommitDate', '')).strip()
  try:
    last_gh_dt = pd.to_datetime(last_gh_date_str)
    current_status = 'Active' if last_gh_dt >= cutoff_date else 'Inactive'
  except Exception:
    current_status = 'Inactive'

  reflection_note = (
      f'{prj} exhibits a {gap_metrics["ActivityPattern"]} development cycle'
      f' with a longest gap of {gap_metrics["LongestGapLength"]} months.'
      f' Commit messages before the gap centered on {before_themes.replace(":", " and ")},'
      f' while post-gap activity indicated {why_recovered.lower()}.'
      f' The project is currently {current_status.lower()} on GitHub.'
  )

  # Write individual reflection markdown file
  refl_filename = f'{netid}_reflection_{prj}.md'
  with open(refl_filename, 'w') as rf:
    rf.write(f'# Reflection: {prj}\n\n')
    rf.write(f'### Inactivity Patterns\n{reflection_note}\n\n')
    rf.write(
        f'- **Longest Inactivity Gap**: {start_str} to {end_str}\n- **Recovery'
        f' Status**: {has_recovered}\n- **Contributors**: {who_recovered}\n-'
        f' **Current Status**: {current_status}\n- **Recent Commit Themes**:'
        f' {recent_themes}\n'
    )

  part3_data.append({
      'project': prj,
      'LongestGapStart (YYYY-MM), ': gap_metrics['LongestGapStart'],
      'LongestGapEnd (YYYY-MM),': gap_metrics['LongestGapEnd'],
      'LongestGapLength': gap_metrics['LongestGapLength'],
      'NumCommitsAfterLastGap': gap_metrics['NumCommitsAfterLastGap'],
      'pattern': gap_metrics['ActivityPattern'],
      'NumberOfGapsInTimeline': gap_metrics['NumberOfGapsInTimeline'],
      'BeforeThemes': before_themes,
      'AfterThemes': after_themes,
      'HypothesizedGapReason': hypo_reason,
      'HasRecovered': has_recovered,
      'WhyRecovered': why_recovered,
      'WhoRecovered': who_recovered,
      'Currentstatus': current_status,
      'RecentThemes': recent_themes,
      'Notes': reflection_note,
  })

# Save jbronyah_project_gap_commits.csv
df_all_gap_commits = pd.concat(gap_commits_list, ignore_index=True)[
    ['pre/post', 'project', 'commit', 'author', 'time', 'message']
]
gap_commits_file = f'{netid}_project_gap_commits.csv'
df_all_gap_commits.to_csv(gap_commits_file, sep=';', index=False)

# Merge new columns into jbronyah_project_stats.csv
df_part3 = pd.DataFrame(part3_data)
# Keep only original Part 1 base columns to avoid duplicate column names
base_cols = [
    'project',
    '#commits',
    '#authors',
    'From',
    'To',
    '#stars',
    '#forks',
    'LastGHCommitDate',
]
df_base = df_stats[[c for c in base_cols if c in df_stats.columns]]
df_updated_stats = df_base.merge(df_part3, on='project', how='left')

# Save updated stats CSV
df_updated_stats.to_csv(stats_file, sep=';', index=False)

print('Generated Deliverables:')
print(f'1. {gap_commits_file} ({len(df_all_gap_commits)} rows)')
print(
    f'2. {stats_file} (successfully updated with all Part 2, 3.1 & 3.2 columns)'
)
print(f'3. 10 Reflection markdown files ({netid}_reflection_<project>.md)')
display(df_updated_stats.head())
```
