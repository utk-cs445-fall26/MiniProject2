```python
import pandas as pd
import matplotlib.pyplot as plt
import requests
```

```python
NETID = "cfinley6"

df = pd.read_csv(f"{NETID}_project_summary.csv", sep=";")
df.columns = ['project','commit','author','time','message']
df['Month'] = pd.to_datetime(df['time'], unit='s').dt.strftime('%Y-%m')
df.head()
```

```python
import matplotlib.ticker as ticker

project_col = 'project'
my_projects = df[project_col].unique()

for prj in my_projects:
    # 1. Filter for the current project
	df1 = df[df[project_col] == prj].copy()
    
	df2 = df1.groupby('Month').size().reset_index(name='Commits')
	monthly_counts = df2
	monthly_counts['time'] = pd.to_datetime(monthly_counts['Month'])

	full_date_range = pd.DataFrame({'time': pd.date_range(start=monthly_counts['time'].min(), end=monthly_counts['time'].max(), freq='MS')})
	
	monthly_counts = pd.merge(full_date_range, monthly_counts, on='time', how='left').fillna(0)
	monthly_counts['Month'] = monthly_counts['time'].dt.strftime('%Y-%m')
	monthly_counts['Commits'] = monthly_counts['Commits'].astype(int)

	plt.figure(figsize=(10,5))
	plt.plot(monthly_counts["Month"], monthly_counts["Commits"], marker="o", linestyle="-")
	plt.xlabel("Time (YYYY-MM)")
	plt.ylabel("Number of Commits")
	plt.title("Commit Activity: "+prj)
	plt.grid(True)
	plt.tight_layout()
	ax =  plt.gca()
	for label in ax.xaxis.get_ticklabels()[::2]:
		label.set_visible(False)
	plt.xticks(rotation=45, fontsize=7)
	plt.savefig(f"{NETID}_timeseries_"+prj+".png", dpi=600)
	plt.show()
	plt.close()
```

# Identify the Longest Inactivity Gap

2022-02 to 2023-04

Add all that info to netid_project_stats.csv

```python
net2prj = pd.read_csv("net2prj.csv")
my_prj_map = net2prj[net2prj['netID'] == NETID].set_index('WoC')

my_projects = df[project_col].unique()
stats_list = []

for prj in my_projects:
    # 1. Get raw project data to calculate basic stats
    df1 = df[df[project_col] == prj].copy()
    
    num_commits = len(df1)
    num_authors = df1['author'].nunique()
    
    # Assuming 'time' in df1 is a unix timestamp or datetime
    if pd.api.types.is_numeric_dtype(df1['time']):
        min_time = pd.to_datetime(df1['time'].min(), unit='s').strftime('%Y-%m-%d')
        max_time = pd.to_datetime(df1['time'].max(), unit='s').strftime('%Y-%m-%d')
    else:
        min_time = pd.to_datetime(df1['time']).min().strftime('%Y-%m-%d')
        max_time = pd.to_datetime(df1['time']).max().strftime('%Y-%m-%d')

    # 2. Get the monthly time series we created in the previous step
    # We will recreate the monthly_counts exactly as before to run the gap logic
    df2 = df1.groupby('Month').size().reset_index(name='Commits')
    monthly_counts = df2
    monthly_counts['time'] = pd.to_datetime(monthly_counts['Month'])
    full_date_range = pd.DataFrame({'time': pd.date_range(start=monthly_counts['time'].min(), end=monthly_counts['time'].max(), freq='MS')})
    monthly_counts = pd.merge(full_date_range, monthly_counts, on='time', how='left').fillna(0)
    monthly_counts['Month'] = monthly_counts['time'].dt.strftime('%Y-%m')
    
    commits = monthly_counts['Commits'].tolist()
    months = monthly_counts['Month'].tolist()

    # 3. Gap Calculation Logic
    max_gap_len = 0
    curr_gap_len = 0
    max_gap_start = "N/A"
    max_gap_end = "N/A"
    curr_gap_start = ""
    gaps_of_3_plus = 0

    for i, c in enumerate(commits):
        if c == 0:
            if curr_gap_len == 0:
                curr_gap_start = months[i]
            curr_gap_len += 1
        else:
            if curr_gap_len >= 3:
                gaps_of_3_plus += 1
            if curr_gap_len > max_gap_len:
                max_gap_len = curr_gap_len
                max_gap_start = curr_gap_start
                max_gap_end = months[i-1]
            curr_gap_len = 0

    # Check if the project ends on a gap
    if curr_gap_len >= 3:
        gaps_of_3_plus += 1
    if curr_gap_len > max_gap_len:
        max_gap_len = curr_gap_len
        max_gap_start = curr_gap_start
        max_gap_end = months[-1]

    # Calculate commits after the longest gap
    if max_gap_len > 0:
        end_idx = months.index(max_gap_end)
        commits_after = sum(commits[end_idx+1:])
    else:
        commits_after = 0
        
    # 4. Determine Activity Pattern (Heuristic - you should manually review your charts to confirm)
    if max_gap_len >= 6 and commits_after > 0:
        pattern = "U-shaped" if gaps_of_3_plus == 1 else "irregular"
    elif max_gap_len >= 3 and commits_after == 0:
        pattern = "declining"
    elif gaps_of_3_plus > 2:
        pattern = "irregular"
    else:
        pattern = "steady" # Change this in the CSV later if it looks rising or cyclical to you!

    # 5. Fetch GitHub Data (Stars, Forks, Last Commit Date)
    gh_url = my_prj_map.loc[prj, 'GH']
    repo_slug = gh_url.replace("https://github.com/", "").strip("/")
    
    try:
        gh_api = f"https://api.github.com/repos/{repo_slug}"
        res = requests.get(gh_api).json()
        stars = res.get('stargazers_count', 0)
        forks = res.get('forks_count', 0)
        
        c_res = requests.get(f"{gh_api}/commits?per_page=1").json()
        last_gh_date = c_res[0]['commit']['committer']['date'][:10] if isinstance(c_res, list) else "N/A"
    except Exception as e:
        stars, forks, last_gh_date = 0, 0, "N/A"

    # Append to stats list
    stats_list.append({
        'project': prj,
        'number of commits': num_commits,
        'number of authors': num_authors,
        'max time': max_time,
        'min time': min_time,
        'number of stars': stars,
        'number of forks': forks,
        'last commit date': last_gh_date,
        'LongestGapStart': max_gap_start,
        'LongestGapEnd': max_gap_end,
        'LongestGapLength': max_gap_len,
        'NumCommitsAfterLastGap': commits_after,
        'NumberOfGapsInTimeline': gaps_of_3_plus,
        'ActivityPattern': pattern
    })

# Save to CSV
stats_df = pd.DataFrame(stats_list)
stats_df.to_csv(f"{NETID}_project_stats.csv", index=False)
print(f"Successfully saved stats for {len(stats_df)} projects to {NETID}_project_stats.csv")
display(stats_df.head()) # Preview the dataframe
```
