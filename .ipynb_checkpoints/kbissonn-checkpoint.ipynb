```python
# Script that looks at df_commit_data.csv to pull the number of commits, number of authors, and min/max time

import pandas as pd
from datetime import datetime, timezone

df = pd.read_csv('df_commit_data.csv')

# group by project and aggregate
summary = df.groupby('project').agg(
    num_commits=('commit', 'nunique'),
    num_authors=('author', 'nunique'),
    max_time=('author_time', 'max'),
    min_time=('author_time', 'min'),
).reset_index()

# convert unix timestamps to readable datetimes
summary['max_time_readable'] = summary['max_time'].apply(
    lambda t: datetime.fromtimestamp(t, tz=timezone.utc).isoformat()
)
summary['min_time_readable'] = summary['min_time'].apply(
    lambda t: datetime.fromtimestamp(t, tz=timezone.utc).isoformat()
)

for _, row in summary.iterrows():
    print(f"{row['project']}: commits={row['num_commits']}, authors={row['num_authors']}, "
          f"max_time={row['max_time_readable']}, min_time={row['min_time_readable']}")
```
