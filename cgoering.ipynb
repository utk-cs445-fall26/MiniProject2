## Step 1: GitHub repo stats

Stars, forks, and last commit date (from `pushed_at`) for each of the 10 repos, straight from the GitHub REST API. This doesn't need auth for a handful of requests, but GitHub rate-limits unauthenticated calls to 60/hour -- if you hit that, set a personal access token in `headers` below.

```python
import requests, time

# Optional: put a GitHub personal access token here to raise the rate limit (60/hr -> 5000/hr)
GITHUB_TOKEN = None  # e.g. 'ghp_xxxxxxxxxxxx'
headers = {'Authorization': f'token {GITHUB_TOKEN}'} if GITHUB_TOKEN else {}

gh_stats = {}
for woc_id, repo in projects.items():
    url = f'https://api.github.com/repos/{repo}'
    r = requests.get(url, headers=headers)
    if r.status_code != 200:
        print(f'ERROR fetching {repo}: {r.status_code} {r.text[:200]}')
        gh_stats[woc_id] = {'stars': None, 'forks': None, 'last_commit_date': None}
        continue
    d = r.json()
    gh_stats[woc_id] = {
        'stars': d.get('stargazers_count'),
        'forks': d.get('forks_count'),
        'last_commit_date': d.get('pushed_at'),  # ISO 8601 timestamp of the most recent push
    }
    time.sleep(0.5)  # be polite to the API

for woc_id, stats in gh_stats.items():
    print(f"{woc_id}: stars={stats['stars']}, forks={stats['forks']}, last_commit_date={stats['last_commit_date']}")
```

## Step 2: WoC commit SHA1s for each project

`get_values('p2c', project)` returns every commit SHA1 WoC has recorded for that project.

```python
from woc.remote import WocMapsRemote
from tqdm import tqdm
import pandas as pd

woc = WocMapsRemote(base_url="https://worldofcode.org/api/")
# if you have an API key:
# woc = WocMapsRemote(base_url="https://worldofcode.org/api/", api_key="woc-XXXXXX-YYYYYY")

list_df_commits = []
for woc_id in projects:
    commits = woc.get_values('p2c', woc_id)
    df_p = pd.DataFrame(commits, columns=["sha1"])
    df_p['project'] = woc_id
    list_df_commits.append(df_p)
    print(f'{woc_id}: {len(df_p)} commit SHA1s')

df_shas = pd.concat(list_df_commits, ignore_index=True)
df_shas.to_csv('df_shas.csv', index=False)  # checkpoint, in case a later step fails
df_shas.head()
```

## Step 3: Full commit data in batches

Batches of 10 SHA1s at a time (the max batch size WoC allows), with a short wait between batches so we don't hammer the API. This runs once per project so a failure partway through doesn't lose everything -- each project's commits get appended to the same running list.

```python
import time

commit_data = []

for woc_id in projects:
    shas = df_shas[df_shas['project'] == woc_id]['sha1']
    chunks = [shas[i:i+10] for i in range(0, len(shas), 10)]

    for chunk in tqdm(chunks, desc=woc_id):
        res, err = woc.get_values_many('commit.tch', chunk.to_list())
        res = {k: v[0] for k, v in res.items()}
        time.sleep(1)  # stay under the rate limit

        if err:
            print(f'Errors for {woc_id}:', err)

        for commit_sha, commit in res.items():
            commit_data.append({
                'project': woc_id,
                'commit': commit_sha,
                'author': commit[2][0],
                'author_time': int(commit[2][1]),
                'message': commit[4],
            })

df_commit_data = pd.DataFrame(commit_data)
df_commit_data.to_csv('df_commit_data.csv', index=False)  # checkpoint
df_commit_data.head()
```

## Step 4: WoC-derived stats per project

Number of commits, number of unique authors, and the earliest/latest commit timestamp -- per project.

```python
woc_stats = df_commit_data.groupby('project').agg(
    ncommits=('commit', 'count'),
    nauthors=('author', 'nunique'),
    max_time=('author_time', 'max'),
    min_time=('author_time', 'min'),
).reset_index()

# also show as human-readable dates for the write-up
woc_stats['max_time_readable'] = pd.to_datetime(woc_stats['max_time'], unit='s')
woc_stats['min_time_readable'] = pd.to_datetime(woc_stats['min_time'], unit='s')
woc_stats
```

### Validation

Sanity check: do the WoC commit counts look reasonable compared to what GitHub shows for each repo (visible on the repo's main page, e.g. "181 Commits")? WoC data is current only up to Jan 2025, so a project with GitHub activity after that will show more commits on GitHub than WoC reports here -- note that per project below once you've checked.

```python
for woc_id in projects:
    row = woc_stats[woc_stats['project'] == woc_id]
    if not row.empty:
        print(f"{woc_id}: WoC reports {row['ncommits'].values[0]} commits, "
              f"{row['nauthors'].values[0]} authors, "
              f"from {row['min_time_readable'].values[0]} to {row['max_time_readable'].values[0]}")
```

## Step 5: Build `cgoering_project_summary.csv`

Semicolon-separated, columns: `project_wocid;commit_sha1;author;time;commit message`. Written once (not appended per project) to avoid duplicate rows on re-run.

```python
df_summary = df_commit_data.rename(columns={
    'project': 'project_wocid',
    'commit': 'commit_sha1',
    'author_time': 'time',
    'message': 'commit message',
})[['project_wocid', 'commit_sha1', 'author', 'time', 'commit message']]

df_summary.to_csv(f'{yournetid}_project_summary.csv', index=False, sep=';')
print(f'Wrote {len(df_summary)} rows to {yournetid}_project_summary.csv')
df_summary.head()
```
